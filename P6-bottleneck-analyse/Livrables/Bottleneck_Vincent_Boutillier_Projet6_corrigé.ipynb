```python
#Afficher les dimensions du dataset
#Consulter le nombre de colonnes
#La nature des données dans chacune des colonnes
#Le nombre de valeurs présentes dans chacune des colonnes
#Afficher les 5 premières lignes de la table
#Vérifier si il y a des lignes en doublon dans la colonne product_id

print("Le tableau comporte {} observation(s) ou article(s)".format(df_erp.shape[0]))
print("Le tableau comporte {} colonne(s)".format(df_erp.shape[1]))

df_erp.dtypes

df_erp.head(5)

duplicates_in_column = df_erp[df_erp['product_id'].duplicated(keep=False)]
if not duplicates_in_column.empty:
    print("Duplicates in column 'product_id':")
    print(duplicates_in_column)
else:
    print("No duplicates found in column 'product_id'.")
```

```python
#Afficher les valeurs distinctes de la colonne stock_status
#À quelle(s) autre(s) colonne(s) sont-elles liées ?
#Création d'une colonne "stock_status_2"
#La valeur de cette deuxième colonne sera fonction de la valeur dans la colonne "stock_quantity"
#Si la valeur de la colonne "stock_quantity" est nulle, renseigner "outofstock" sinon mettre "instock"

# ===== ANALYSE stock_status + CORRÉLATIONS =====
print("🔍 VALEURS UNIQUES stock_status")
print("=" * 50)

# 1. VALEURS DISTINCTES de stock_status
print("📊 VALEURS UNIQUES :")
valeurs_uniques = erp['stock_status'].unique()
print(f"Nombre de valeurs uniques : {len(valeurs_uniques)}")
print("Valeurs :", list(valeurs_uniques))

print(f"\n📈 RÉPARTITION par stock_status :")
repartition = erp['stock_status'].value_counts()
print(repartition)

# 2. LIENS avec les AUTRES COLONNES (groupement)
print(f"\n🔗 CORRÉLATIONS avec AUTRES COLONNES :")
print("-" * 60)

colonnes_analyse = ['onsale_web', 'price', 'stock_quantity', 'purchase_price']
for col in colonnes_analyse:
    print(f"\n{col.upper()} par stock_status :")
    grouped = erp.groupby('stock_status')[col].agg(['count', 'mean', 'min', 'max']).round(2)
    display(grouped)

# 3. VISUALISATION CROISÉE (stock_quantity vs stock_status)
print(f"\n📊 CROISÉ stock_quantity & stock_status :")
stock_croise = erp.groupby('stock_status')['stock_quantity'].agg(['count', 'mean', 'sum']).round(2)
print("Nombre | Stock moyen | Stock total")
display(stock_croise)

# 4. EXEMPLES CONCRETS (5 premières lignes par statut)
print(f"\n📋 EXEMPLES par stock_status :")
for status in erp['stock_status'].unique():
    print(f"\n{status.upper()} (5 premiers exemples) :")
    exemple = erp[erp['stock_status'] == status][['product_id', 'price', 'stock_quantity', 'purchase_price']].head()
    display(exemple)



```

```python
# ===== DIAGNOSTIC PROBLÈMES ERP =====
print("🚨 DIAGNOSTIC DES ANOMALIES ERP")
print("=" * 50)

# 1. STOCKS NÉGATIFS (CRITIQUE)
stocks_negatifs = erp[erp['stock_quantity'] < 0]
print(f"❌ STOCKS NÉGATIFS : {len(stocks_negatifs)} produits")
print("Détails :")
display(stocks_negatifs[['product_id', 'stock_status', 'stock_quantity', 'price']].sort_values('stock_quantity'))

# 2. PRODUITS outofstock AVEC stock > 0
outofstock_positif = erp[(erp['stock_status'] == 'outofstock') & (erp['stock_quantity'] > 0)]
print(f"\n❌ OUTOFSTOCK mais stock > 0 : {len(outofstock_positif)} produits")
display(outofstock_positif[['product_id', 'stock_status', 'stock_quantity']])

# 3. INSTOCK mais stock = 0
instock_zero = erp[(erp['stock_status'] == 'instock') & (erp['stock_quantity'] == 0)]
print(f"\n⚠️  INSTOCK mais stock = 0 : {len(instock_zero)} produits")

# 4. PRIX NÉGATIFS
prix_negatifs = erp[erp['price'] < 0]
print(f"\n💰 PRIX NÉGATIFS : {len(prix_negatifs)} produits")
if len(prix_negatifs) > 0:
    display(prix_negatifs[['product_id', 'price', 'purchase_price']])
```

```python
# ===== CRÉATION stock_status_2 =====
print("🔧 CRÉATION stock_status_2 (basée sur stock_quantity)")
print("=" * 50)

# Création de la colonne stock_status_2
# Règle : stock_quantity == 0 → "outofstock", sinon → "instock"
erp['stock_status_2'] = erp['stock_quantity'].apply(
    lambda x: 'outofstock' if x == 0 else 'instock'
)

# Vérification de la nouvelle colonne
print("✅ Colonne stock_status_2 créée !")
print(f"Type : {erp['stock_status_2'].dtype}")
print(f"Valeurs uniques : {erp['stock_status_2'].nunique()}")

# Répartition
print("\n📊 RÉPARTITION stock_status_2 :")
print(erp['stock_status_2'].value_counts())

# COMPARAISON stock_status vs stock_status_2
print(f"\n🔍 COMPARAISON des 2 colonnes :")
comparaison = erp.groupby(['stock_status', 'stock_status_2']).size().unstack(fill_value=0)
display(comparaison)

# INCOHÉRENCES détectées
incoherents = erp[erp['stock_status'] != erp['stock_status_2']]
print(f"\n🚨 INCOHÉRENCES détectées : {len(incoherents)} produits")
print("Exemples :")
display(incoherents[['product_id', 'stock_status', 'stock_status_2', 'stock_quantity']].head())
```

```python
#Mais il est possible de synthétiser ce résultat en effectuant la somme de cette colonne:
#True vaut 1 et False 0
#Nous devrions obtenir la somme de 825 qui correspond au nombre de lignes dans ce dataset

# ===== AFFINEMENT : TRUE=1, FALSE=0 =====
print("🔢 CONVERSION BOOLÉEN → NUMÉRIQUE (1/0)")
print("=" * 50)

# Conversion True → 1, False → 0
comparaison_numerique = (erp['stock_status'] == erp['stock_status_2']).astype(int)
print("✅ Conversion réussie ! Colonne numérique créée")

# STATISTIQUES
print(f"\n📊 STATISTIQUES 1/0 :")
print(f"Somme (nb 1)     : {comparaison_numerique.sum()}")
print(f"Somme (nb 0)     : {len(comparaison_numerique) - comparaison_numerique.sum()}")
print(f"Moyenne (taux 1) : {comparaison_numerique.mean():.3f}")
print(f"Taux d'erreur    : {(1 - comparaison_numerique.mean())*100:.1f}%")

# AFFICHAGE
print("\n📋 10 PREMIÈRES VALEURS (1/0) :")
display(comparaison_numerique.head(10))

# DÉTAIL DES ZÉROS (erreurs)
erreurs = erp[comparaison_numerique == 0]
print(f"\n🚨 {len(erreurs)} ERREURS (valeur = 0) :")
display(erreurs[['product_id', 'stock_status', 'stock_status_2', 'stock_quantity']])

# SAUVEGARDE dans ERP
erp['comparaison_1_0'] = comparaison_numerique
print("\n✅ Colonne 'comparaison_1_0' ajoutée au dataset")
```

```python
#Calculer la moyenne du prix

#Calculer l'écart-type du prix

#Calculer le Z-score


moyenne_prix = df_final['price'].mean()
print(f"📊 MOYENNE des prix : {moyenne_prix:.2f}€")

ecart_type_prix = df_final['price'].std()
print(f"📏 ÉCART-TYPE des prix : {ecart_type_prix:.2f}€")

df_final['z_score'] = (df_final['price'] - moyenne_prix) / ecart_type_prix
print("\n✅ Z-SCORE calculé pour tous les produits !")
print("Exemple (5 premiers) :")
print(df_final[['price', 'z_score']].head().round(2))
```

```python
#Quel est le seuil prix dont le z-score est supérieur à 3?

moyenne = df_final['price'].mean()
ecart_type = df_final['price'].std()

# Z-Score supérieur à 2
seuil_z3 = moyenne + 2 * ecart_type

print(f"📊 Moyenne : {moyenne:.2f}€")
print(f"📏 Écart-type : {ecart_type:.2f}€")
print(f"🚨 SEUIL Z=2 : {seuil_z3:.2f}€")
print(f"💰 Prix au-dessus : {len(df_final[df_final['price'] > seuil_z3])} produits")
```

<h3 style="margin: auto; padding: 20px; color: RGB(51,165,182); ">Etape 4.2.2 - Identification par l'intervalle interquartile</h3>
</div>

```python
#Utilisation de la fonction "describe" de Pandas pour l'étude des mesures de dispersion
#Définir le nombre d'articles et la proportion de l'ensemble du catalogue "outliers"


print("📈 MESURES DE DISPERSION")
stats = df_final['price'].describe()
print(stats.round(2))

# Q1 et Q3  depuis describe()
Q1 = stats['25%']
Q3 = stats['75%']
IQR = Q3 - Q1

print(f"\n📏 Intervalle InterQuartile (IQR) :")
print(f"Q1 (25%) : {Q1:.2f}€")
print(f"Q3 (75%) : {Q3:.2f}€")
print(f"IQR (Q3-Q1) : {IQR:.2f}€")

# Règle IQR : outlier si > Q3 + 1.5×IQR
seuil_superieur = Q3 + 1.5 * IQR
seuil_inferieur = Q1 - 1.5 * IQR

print(f"\n🔍 SEUILS OUTLIERS Intervalle InterQuartile :")
print(f"Seuil SUP : {seuil_superieur:.2f}€")
print(f"Seuil INF : {seuil_inferieur:.2f}€")

# Nb d'outliers
outliers_iqr = df_final[(df_final['price'] > seuil_superieur) | (df_final['price'] < seuil_inferieur)]
print(f"🚨 {len(outliers_iqr)} OUTLIERS détectés par IQR")
```

```python
#Définir un seuil pour les articles "outliers" en prix

# Seuil
Q1 = df_final['price'].quantile(0.25)
Q3 = df_final['price'].quantile(0.75)
seuil_outlier = Q3 + 1.5 * (Q3 - Q1)

print(f" SEUIL OUTLIER = {seuil_outlier:.0f}€")
print(f" Tout > {seuil_outlier:.0f}€ = Produits 'Premium'")
```

```python
#Selon vous, ces outliers sont-ils justifiés ? Comment le démontrer si cela est possible ?

#Oui c'est justifié car en général ce sont des Alcool forts ou des grands crus (Champagne, Cognac, Gin, Whisky)


Q1 = df_final['price'].quantile(0.25)
Q3 = df_final['price'].quantile(0.75)

IQR = Q3 - Q1
outliers = df_final[(df_final['price'] < (Q1 - 1.5 * IQR)) | (df_final['price'] > (Q3 + 1.5 * IQR))]

print(f'Les articles considérés comme des outliers prix : {outliers}')
```

```python
#############################
# Calculer le 20 / 80 en CA #
#############################

#Créer une colonne calculant la part du CA de la ligne dans le dataset
#Créer une colonne réalisant la somme cumulative de la colonne précedemment créée
#Grâce aux deux colonnes créées précedemment, calculer le nombre d'articles représentant 80% du CA
#Afficher la proportion que représente ce groupe d'articles dans le catalogue entier du site web


import numpy as np

# 1. Nettoyer CA (négatifs → 0)
df_ca = df_final.copy()  # Garde tes 8469 lignes
df_ca['ca_par_article_clean'] = np.maximum(0, df_ca['ca_par_article'])

# 2. TRIER CA décroissant
df_ca = df_ca.sort_values('ca_par_article_clean', ascending=False).reset_index(drop=True)

# 3. % + cumulatif
df_ca['%_ca_par_article'] = df_ca['ca_par_article_clean'] / df_ca['ca_par_article_clean'].sum()
df_ca['somme_cumulative'] = df_ca['%_ca_par_article'].cumsum()

# 4. 80% CA
articles_80_ca = df_ca[df_ca['somme_cumulative'] <= 0.80]
nombre_articles_80_ca = len(articles_80_ca)

# 5. Proportion sur catalogue entier (8469)
proportion_articles_80_ca = nombre_articles_80_ca / len(df_final)

print(f"🎯 80/20 CA (sur {len(df_final)} articles)")
print(f"{nombre_articles_80_ca} articles = 80% CA")
print(f"Proportion : {proportion_articles_80_ca:.1%}")
print(f"({nombre_articles_80_ca}/8469 articles)")
```

```python
############################
# Analyse du taux de marge #
############################

#Création de la colonne Prix HT (Sachant que la valeur de la TVA en 2026 sur les produits (dont l'alcool) est de 20% donc 1.2)
#Création de la colonne Taux de marge
#Afficher le prix minimum de la colonne "taux_marge"
#Afficher le prix maximum de la colonne "taux_marge"

df_final["Prix_HT"] = df_final["price"] / (1+0.2)
df_final["Prix_achat_HT"] = df_final["purchase_price"] / (1+0.2)
df_final["taux_marge"] = (df_final["Prix_HT"] - df_final["Prix_achat_HT"]) / df_final["Prix_achat_HT"]

print("Le taux de marge le plus bas: {}".format(df_final["taux_marge"].min()))
print("Le taux de marge le plus haut: {}".format(df_final["taux_marge"].max()))
```

```python
#Création d'un dataframe avec les taux positifs
#Afficher le prix minimum de la colonne "taux_marge"
#Afficher le prix maximum de la colonne "taux_marge"

df_tx_positif = df_final[df_final['taux_marge'] >= 0]
print("Le taux de marge le plus bas: {}".format(df_tx_positif["taux_marge"].min()))
print("Le taux de marge le plus haut: {}".format(df_tx_positif["taux_marge"].max()))
```
